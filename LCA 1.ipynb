```python
my_dict = {"name": "Aryan","age": 19,"course": "Computer Science"}
print("Original Dictionary:", my_dict)
```

```python
my_dict = {"name": "Aryan","age": 19,"course": "Computer Science"}
my_dict["college"] = "MIT-WPU"
print("After Adding:", my_dict)
```

```python
my_dict = {"name": "Aryan","age": 19,"course": "Computer Science","college": "MIT-WPU"}
my_dict.pop("course")
print("After Removing Course:",my_dict)
```
